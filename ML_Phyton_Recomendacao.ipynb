```python
import pandas as pd
```

```python
filmes = pd.read_csv('/content/movies.csv')
filmes.columns = ['filmeId', 'titulo', 'generos']
filmes = filmes.set_index('filmeId')
filmes.head()
```

```python
notas = pd.read_csv('/content/ratings.csv')
notas.columns = ['usuarioId', 'filmeId', 'nota', 'momento']
notas.head()
```

```python
notas.describe()
```

# Primeira tentativa de recomendação: heurística de total de votos






```python
#Descobrindo a frequência com que cada filme foi visto
total_de_votos = notas['filmeId'].value_counts()
total_de_votos.head()
```

```python
filmes.loc[318]
```

```python
# Ornenando os mais votados e recomenando os primeiros
filmes['total_de_votos'] = total_de_votos
filmes.head()
```

```python
filmes.sort_values('total_de_votos', ascending=False).head(10)
```

```python
notas_medias = notas.groupby('filmeId').mean()['nota']
notas_medias.head()
```

```python
filmes['nota_media'] = notas_medias
filmes.sort_values('total_de_votos', ascending=False).head(10)
```

Segunda heurística: nota média e filtrando votos


```python
# Ordenando pela nota média
filmes.sort_values('nota_media', ascending=False).head(10)
```

```python
filmes.query('total_de_votos >= 10').sort_values('nota_media', ascending=False).head(10)
```

```python
filmes_com_mais_de_50_votos = filmes.query('total_de_votos >= 50')
filmes_com_mais_de_50_votos.sort_values('nota_media', ascending=False).head(10)
```

Outra possibilidade de recomendação, agora levando em consideração os filmes que o usuário assistiu

```python
eu_assisti = [1, 21, 19, 10, 11, 7, 2]
filmes.loc[eu_assisti]
```

```python
aventura_infantil_e_fantasia = filmes_com_mais_de_50_votos.query("generos=='Adventure|Children|Fantasy'")
aventura_infantil_e_fantasia.drop(eu_assisti, errors='ignore').sort_values('nota_media', ascending=False).head(10)
```

Colaboração dos usuários x informações baseadas em conteúdo

```python
# Procurando usuários similares
def notas_do_usuario(usuario):
  notas_do_usuario = notas.query('usuarioId==%d' % usuario)
  notas_do_usuario = notas_do_usuario[['filmeId','nota']].set_index('filmeId')
  return notas_do_usuario

```

```python
usuario1 = notas_do_usuario(1)
usuario4 = notas_do_usuario(4)
```

```python
usuario1.head()
```

```python
usuario4.head()
```

Distância entre usuários do dataset


```python
import numpy as np
```

```python
# Calculando a distancia de dois vetores.
def distancia_de_vetores(a, b):
  return np.linalg.norm(a - b)
```

```python
# Juntando dois dataframes, onde será apresentando os que os dois usários assistiram em comum e também a distancia entre o 1 e o 4.
# Obs: aparece apenas os filmes do usuário a esquerda, sendo procurado pelo índice.
diferencas = usuario1.join(usuario4, lsuffix='_esquerda', rsuffix='r_direita').dropna()
distancia_de_vetores(diferencas['nota_esquerda'], diferencas['notar_direita'])
```

```python
# Transformando em uma função.
def distancia_de_usuarios(usuario_id1, usuario_id2):
  notas1 = notas_do_usuario(usuario_id1)
  notas2 = notas_do_usuario(usuario_id2)
  diferencas = notas1.join(notas2, lsuffix='_esquerda', rsuffix='r_direita').dropna()
  return distancia_de_vetores(diferencas['nota_esquerda'], diferencas['notar_direita'])
```

```python
distancia_de_usuarios(1, 4)
```
